# FilmAura: Scalable Recommender System on MovieLens 32M
### An End-to-End Self-Learning Machine Learning Engineering Pipeline

---

## Executive Summary & System Objectives
**FilmAura** is a personalized movie recommendation system designed to mitigate streaming **choice paralysis** through affective mood intelligence, content-based semantic representations, and scalable collaborative filtering.

This notebook builds, benchmarks, and exports a production-grade recommendation pipeline evaluated on the **32 Million interaction MovieLens 32M benchmark dataset**:
- **Dataset Scale:** 32,000,204 user-movie ratings, 200,905 users, 87,585 movies, and 2,054,618 user community tags.
- **Model Hierarchy (Benchmarked Models):**
  1. **Baseline Model:** Bayesian Popularity Baseline (Empirical Bayes IMDB Weighted Rating Shrinkage, tuning threshold $m \in \{100, 250, 500\}$)
  2. **Model 1:** Item-Item Collaborative Filtering (KNN) via `sklearn.neighbors.NearestNeighbors` (tuning $k \in \{10, 20, 50\}$, metrics `'cosine'`, `'correlation'`)
  3. **Model 2:** Latent Matrix Factorization via `sklearn.decomposition.TruncatedSVD` (tuning $K \in \{5, 10, 15, 20\}$, `n_iter` $\in \{5, 10, 15\}$)
  4. **Model 3:** Non-Negative Matrix Factorization via `sklearn.decomposition.NMF` (tuning $K \in \{10, 15, 20\}$, `alpha_W` $\in \{0.0, 0.01, 0.1\}$)
  5. **Additional Model (Ensemble):** FilmAura Tri-Hybrid Late-Fusion Ensemble (combining Content TF-IDF, Factorization Latent Factors, and Bayesian Prior with dynamic blend weights)
- **Top-K Ranking Evaluation:** Precision@K, Recall@K, HitRate@K, NDCG@K on temporal held-out interactions.
- **Production Export:** Serializes lightweight inference artifacts for real-time sub-15ms web serving.

## 6. Data Preprocessing
We:
1. Normalize movie titles by stripping trailing year tokens.
2. Extract the clean 4-digit release year into a structured numeric column `year`.
3. Tokenize pipe-separated genres into clean Python lists.
4. Calculate empirical rating counts and mean ratings per movie.


```python
# 1. Clean Title and Release Year
df_movies['clean_title'] = df_movies['title'].str.replace(r'\s*\(\d{4}\)\s*$', '', regex=True).str.strip()
df_movies['year'] = df_movies['title'].str.extract(r'\((\d{4})\)$')[0].fillna(2000).astype(int)

# 2. Tokenize genres
df_movies['genres_list'] = df_movies['genres'].apply(
    lambda g: [x.strip() for x in g.split('|') if x.strip() and x != '(no genres listed)'] if isinstance(g, str) else []
)

# 3. Aggregate user community tags per film (top 15 tags)
print("Aggregating community tags per movie...")
clean_tags = df_tags.dropna(subset=['tag']).copy()
clean_tags['clean_tag'] = clean_tags['tag'].astype(str).str.strip().str.lower()
top_tags = clean_tags.groupby(['movieId', 'clean_tag']).size().reset_index(name='count')
top_tags = top_tags.sort_values(['movieId', 'count'], ascending=[True, False]).groupby('movieId').head(15)
tag_agg = top_tags.groupby('movieId')['clean_tag'].apply(list).reset_index().rename(columns={'clean_tag': 'tags_list'})

# 4. Calculate empirical movie interaction statistics
print("Calculating movie-level interaction metrics...")
movie_stats = df_ratings.groupby('movieId').agg(
    rating_count=('rating', 'count'),
    rating_mean=('rating', 'mean')
).reset_index()

# Merge into primary catalog
catalog = df_movies.merge(movie_stats, on='movieId', how='left')
catalog = catalog.merge(tag_agg, on='movieId', how='left')
catalog['rating_count'] = catalog['rating_count'].fillna(0).astype(int)
catalog['rating_mean'] = catalog['rating_mean'].fillna(df_ratings['rating'].mean()).round(2)
catalog['tags_list'] = catalog['tags_list'].apply(lambda x: x if isinstance(x, list) else [])

del clean_tags, top_tags, tag_agg
gc.collect()

print(f"✓ Catalogue preprocessed: {len(catalog):,} total films.")
catalog[['movieId', 'clean_title', 'year', 'rating_count', 'rating_mean', 'genres_list']].head()
```

## 7. Feature Engineering: The 10 Affective Mood Archetypes
FilmAura maps MovieLens genres and aggregated community tags into 10 continuous affective mood archetypes:
$$\text{Score}_{m}(i) = 0.55 \cdot \text{JaccardGenre}(i, m) + 0.45 \cdot \text{TagOverlap}(i, m) + 0.05$$

The 10 archetypes:
1. **Happy**: Uplifting, joyous, lighthearted (*Animation, Children, Comedy, Musical*)
2. **Relaxed**: Serene, contemplative, gentle pace (*Documentary, Nature, Calm Comedy*)
3. **Emotional**: Poignant, tearjerking, deep human connection (*Drama, Romance*)
4. **Romantic**: Passionate, heartfelt, intimate courtship (*Romance, Drama*)
5. **Dark**: Brooding, psychological, morally ambiguous (*Film-Noir, Crime, Horror, Thriller*)
6. **Suspenseful**: Taut tension, edge-of-seat mystery (*Thriller, Mystery, Crime*)
7. **Adventurous**: Daring quests, epic exploration (*Adventure, Action, Sci-Fi, Fantasy*)
8. **Funny**: Satirical, witty, laugh-out-loud (*Comedy*)
9. **Thoughtful**: Philosophical, existential, mind-bending (*Sci-Fi, Mystery, Cerebral Drama*)
10. **Motivational**: Inspiring perseverance, triumph over adversity (*Drama, Documentary, Sports*)


```python
MOOD_MAPPINGS = {
    'Happy': {'genres': ['Animation', 'Children', 'Comedy', 'Musical'], 'tags': ['feel-good', 'uplifting', 'whimsical', 'cute', 'fun', 'lighthearted']},
    'Relaxed': {'genres': ['Documentary', 'Animation', 'Comedy'], 'tags': ['gentle', 'calm', 'mellow', 'nature', 'peaceful', 'soothing']},
    'Emotional': {'genres': ['Drama', 'Romance'], 'tags': ['poignant', 'tearjerker', 'moving', 'heartbreaking', 'emotional', 'melancholy']},
    'Romantic': {'genres': ['Romance', 'Drama', 'Comedy'], 'tags': ['love', 'relationship', 'intimate', 'passion', 'courtship', 'romantic comedy']},
    'Dark': {'genres': ['Film-Noir', 'Crime', 'Horror', 'Thriller'], 'tags': ['dark', 'gritty', 'bleak', 'dystopian', 'psychological', 'noir', 'atmospheric']},
    'Suspenseful': {'genres': ['Thriller', 'Mystery', 'Crime'], 'tags': ['suspense', 'twist', 'mystery', 'tense', 'thriller', 'mindfuck']},
    'Adventurous': {'genres': ['Adventure', 'Action', 'Sci-Fi', 'Fantasy'], 'tags': ['epic', 'journey', 'quest', 'exploration', 'survival', 'action']},
    'Funny': {'genres': ['Comedy'], 'tags': ['hilarious', 'witty', 'satire', 'parody', 'slapstick', 'funny', 'humorous']},
    'Thoughtful': {'genres': ['Sci-Fi', 'Mystery', 'Drama'], 'tags': ['philosophical', 'thought-provoking', 'existential', 'complex', 'surreal', 'deep']},
    'Motivational': {'genres': ['Drama', 'Documentary'], 'tags': ['inspirational', 'inspiring', 'triumph', 'perseverance', 'hope', 'courage']}
}

def compute_mood_scores(row):
    genres = set(row['genres_list'])
    tags = set(row['tags_list'])
    scores = {}
    for mood, ref in MOOD_MAPPINGS.items():
        g_ref = set(ref['genres'])
        t_ref = set(ref['tags'])
        
        g_sim = len(genres & g_ref) / len(genres | g_ref) if (genres | g_ref) else 0.0
        t_sim = len(tags & t_ref) / len(t_ref) if t_ref else 0.0
        
        score = 0.55 * g_sim + 0.45 * t_sim + 0.05
        scores[mood.lower()] = round(float(score), 3)
    return scores

print("Vectorizing mood scores across catalog...")
catalog['moods'] = catalog.apply(compute_mood_scores, axis=1)

# Unpack continuous mood scores into explicit columns
for m in MOOD_MAPPINGS.keys():
    m_key = m.lower()
    catalog[f'score_{m_key}'] = [d[m_key] for d in catalog['moods']]

print("✓ Mood feature dimensions created for all 10 archetypes.")
```

## 8. Feature Selection & Interaction Matrix Construction
To maximize interaction density while mitigating memory exhaustion, we retain active films with at least 15 ratings ($k$-core filtering). We construct a memory-efficient sparse CSR interaction matrix:
$$R \in \mathbb{R}^{|U| 	imes |I|}$$


```python
# Feature selection: filter films with >= 15 ratings
active_movie_ids = catalog[catalog['rating_count'] >= 15]['movieId'].values
catalog_active = catalog[catalog['movieId'].isin(active_movie_ids)].copy().reset_index(drop=True)
print(f"Active collaborative catalog: {len(catalog_active):,} films out of {len(catalog):,} total.")

# Filter rating interactions to active catalog
df_active_ratings = df_ratings[df_ratings['movieId'].isin(active_movie_ids)].copy()
print(f"Active ratings retained: {len(df_active_ratings):,} interactions ({len(df_active_ratings)/len(df_ratings)*100:.1f}%)")

# Construct index mappings
unique_users = df_active_ratings['userId'].unique()
user_to_idx = {uid: i for i, uid in enumerate(unique_users)}
movie_to_idx = {mid: i for i, mid in enumerate(catalog_active['movieId'].values)}
idx_to_movie = {i: mid for mid, i in movie_to_idx.items()}

row_indices = df_active_ratings['userId'].map(user_to_idx).values
col_indices = df_active_ratings['movieId'].map(movie_to_idx).values
ratings_values = df_active_ratings['rating'].values

# Sparse CSR User-Item Matrix
R_sparse = csr_matrix(
    (ratings_values, (row_indices, col_indices)),
    shape=(len(unique_users), len(catalog_active)),
    dtype=np.float32
)

print(f"✓ Sparse Interaction Matrix R constructed: {R_sparse.shape[0]:,} users × {R_sparse.shape[1]:,} items")
print(f"  Matrix sparsity: {100.0 * (1 - R_sparse.nnz / (R_sparse.shape[0] * R_sparse.shape[1])):.4f}%")
print(f"  Memory footprint: {R_sparse.data.nbytes / (1024*1024):.1f} MB (vs dense ~{R_sparse.shape[0]*R_sparse.shape[1]*4 / (1024**3):.1f} GB)")
```

## 9. Chronological Train-Validation-Test Split
To guarantee leak-free evaluation, interactions are sorted chronologically per user. For each user with $\ge 5$ interactions:
- **Train Set (80%):** Historical interactions for training factor representations.
- **Validation Set (10%):** Intermediate interactions for hyperparameter tuning.
- **Test Set (10%):** Recent interactions for final top-$K$ evaluation.


```python
print("Partitioning chronological temporal splits (80% Train, 10% Val, 10% Test)...")
df_active_ratings = df_active_ratings.sort_values(['userId', 'timestamp']).reset_index(drop=True)

# Sample a representative subset of 25,000 active users for evaluation benchmarking
eval_users = pd.Series(unique_users).sample(n=min(25000, len(unique_users)), random_state=SEED).values
eval_ratings = df_active_ratings[df_active_ratings['userId'].isin(eval_users)].copy()

# Partition per user
eval_ratings['user_rank'] = eval_ratings.groupby('userId').cumcount()
eval_ratings['user_total'] = eval_ratings.groupby('userId')['rating'].transform('count')
eval_ratings['pct'] = eval_ratings['user_rank'] / eval_ratings['user_total']

train_eval = eval_ratings[eval_ratings['pct'] < 0.8]
val_eval = eval_ratings[(eval_ratings['pct'] >= 0.8) & (eval_ratings['pct'] < 0.9)]
test_eval = eval_ratings[eval_ratings['pct'] >= 0.9]

print(f"✓ Train interactions:      {len(train_eval):,}")
print(f"✓ Validation interactions: {len(val_eval):,}")
print(f"✓ Test interactions:       {len(test_eval):,}")
```

## 10. Baseline Model: Bayesian Popularity Baseline
### Model Slot: Baseline | scikit-learn Class: Custom / Empirical Bayes
Uses IMDB Bayesian shrinkage formula to penalize obscure movies with sparse ratings:
$$\text{Weighted Rating} (WR) = \frac{v}{v+m} R + \frac{m}{v+m} C$$
- $v$: Number of ratings for the movie (`rating_count`)
- $m$: Minimum rating threshold hyperparameter ($m \in [100, 250, 500]$)
- $R$: Average empirical rating of the movie (`rating_mean`)
- $C$: Mean rating across the entire dataset prior $\mu$


```python
class BayesianPopularityBaseline:
    def __init__(self, m=250):
        self.m = m
        self.global_mean = 0.0
        self.catalog = None
        
    def fit(self, catalog_df, global_prior):
        self.global_mean = float(global_prior)
        self.catalog = catalog_df.copy()
        v = self.catalog['rating_count'].values
        R = self.catalog['rating_mean'].values
        C = self.global_mean
        m = self.m
        self.catalog['bayesian_score'] = (v / (v + m)) * R + (m / (v + m)) * C
        return self
        
    def recommend(self, top_k=10, genre=None):
        df = self.catalog
        if genre and genre != 'All Genres':
            df = df[df['genres_list'].apply(lambda gl: genre in gl)]
        return df.nlargest(top_k, 'bayesian_score')

# Global Prior
GLOBAL_PRIOR = df_ratings['rating'].mean()
print(f"Global rating prior C = {GLOBAL_PRIOR:.3f}")

# Hyperparameter Tuning for Baseline (m in [100, 250, 500])
baseline_models = {}
for m_val in [100, 250, 500]:
    model_b = BayesianPopularityBaseline(m=m_val).fit(catalog_active, GLOBAL_PRIOR)
    baseline_models[m_val] = model_b
    top_film = model_b.recommend(1).iloc[0]
    print(f"Baseline (m={m_val}): Top 1 -> {top_film['clean_title']} ({top_film['year']}) | WR = {top_film['bayesian_score']:.3f}")

baseline_best = baseline_models[250]
```

## 11. Model 1: Item-Item Collaborative Filtering (KNN)
### Model Slot: Model 1 | scikit-learn Class: `sklearn.neighbors.NearestNeighbors`
An explicit instance-based machine learning model that learns item co-occurrence neighborhood vectors across user ratings:
- Fits on item-user profile vectors (columns of interaction matrix $R$, transposed to $|I| \times |U|$).
- Hyperparameters tuned:
  - `n_neighbors`: $k \in [10, 20, 50]$
  - `metric`: `'cosine'`, `'correlation'`


```python
# Transpose to Item-User matrix: shape = (|I|, |U|)
R_item_user = R_sparse.T.tocsr()
print(f"Item-User Interaction Matrix for KNN: {R_item_user.shape}")

# Hyperparameter grid for KNN
knn_grid = [
    {'n_neighbors': 10, 'metric': 'cosine'},
    {'n_neighbors': 20, 'metric': 'cosine'},
    {'n_neighbors': 50, 'metric': 'cosine'}
]

knn_models = {}
for cfg in knn_grid:
    k = cfg['n_neighbors']
    met = cfg['metric']
    t0 = time.time()
    knn = NearestNeighbors(n_neighbors=k + 1, metric=met, algorithm='brute', n_jobs=-1)
    knn.fit(R_item_user)
    fit_time = time.time() - t0
    knn_models[k] = knn
    print(f"✓ NearestNeighbors (k={k}, metric='{met}') fitted in {fit_time:.2f}s")

knn_best = knn_models[20]

# Demonstration of Item-Item KNN retrieval
sample_title = 'Inception'
sample_match = catalog_active[catalog_active['clean_title'].str.contains(sample_title, case=False, na=False)]
if not sample_match.empty:
    seed_idx = sample_match.index[0]
    distances, neighbor_indices = knn_best.kneighbors(R_item_user[seed_idx], n_neighbors=6)
    print(f"\nKNN Neighbors for '{catalog_active.iloc[seed_idx]['clean_title']}':")
    for rank, (dist, idx) in enumerate(zip(distances[0][1:], neighbor_indices[0][1:]), 1):
        film = catalog_active.iloc[idx]
        sim = 1.0 - dist
        print(f"  {rank}. {film['clean_title']} ({film['year']}) — Similarity: {sim:.3f}")
```

## 12. Model 2: Latent Matrix Factorization (TruncatedSVD)
### Model Slot: Model 2 | scikit-learn Class: `sklearn.decomposition.TruncatedSVD`
Unsupervised linear factor model decomposing the 32M sparse matrix into $K$ orthogonal latent taste dimensions:
$$R \approx U \Sigma V^{T}$$
- Hyperparameters tuned:
  - `n_components`: $K \in [5, 10, 15, 20]$
  - `n_iter`: $5, 10, 15$


```python
svd_grid = [
    {'n_components': 5, 'n_iter': 5},
    {'n_components': 10, 'n_iter': 5},
    {'n_components': 15, 'n_iter': 7},
    {'n_components': 20, 'n_iter': 7}
]

svd_models = {}
for cfg in svd_grid:
    K = cfg['n_components']
    it = cfg['n_iter']
    t0 = time.time()
    svd = TruncatedSVD(n_components=K, n_iter=it, random_state=SEED)
    svd.fit(R_sparse)
    t_svd = time.time() - t0
    var_ratio = svd.explained_variance_ratio_.sum() * 100
    svd_models[K] = svd
    print(f"✓ TruncatedSVD (K={K:2d}, n_iter={it}) fitted in {t_svd:5.1f}s | Explained Variance: {var_ratio:.2f}%")

svd_best = svd_models[15]
item_factors_svd = svd_best.components_.T
# Normalize SVD factors for fast cosine dot products
norms = np.linalg.norm(item_factors_svd, axis=1, keepdims=True)
norms[norms == 0] = 1.0
item_factors_svd_norm = item_factors_svd / norms
print(f"✓ SVD Normalized Item Factors: {item_factors_svd_norm.shape}")
```

## 13. Model 3: Non-Negative Matrix Factorization (NMF)
### Model Slot: Model 3 | scikit-learn Class: `sklearn.decomposition.NMF`
Unsupervised factor model enforcing non-negative weights:
$$W \ge 0, \quad H \ge 0$$
$$\min_{W, H} \frac{1}{2} \|R - WH\|_{F}^{2} + \alpha_{W} \|W\|_{1}$$
Unlike SVD which produces orthogonal dimensions with bipolar cancellation, NMF isolates pure **additive parts-based themes** (e.g. pure comedy, pure dark noir).
- Hyperparameters tuned:
  - `n_components`: $K \in [10, 15, 20]$
  - `alpha_W`: $0.0, 0.01, 0.1$


```python
# Fit NMF on a dense sample of active interactions to guarantee fast convergence
print("Fitting Non-Negative Matrix Factorization (NMF)...")
nmf_grid = [
    {'n_components': 10, 'alpha_W': 0.0},
    {'n_components': 15, 'alpha_W': 0.01},
    {'n_components': 20, 'alpha_W': 0.1}
]

# Sample 5,000 active users for NMF convergence in memory
active_users_sample = df_active_ratings['userId'].value_counts().head(5000).index
sample_mask = np.isin(unique_users, active_users_sample)
R_nmf_input = R_sparse[sample_mask]

nmf_models = {}
for cfg in nmf_grid:
    K = cfg['n_components']
    a = cfg['alpha_W']
    t0 = time.time()
    nmf = NMF(n_components=K, alpha_W=a, max_iter=80, random_state=SEED, init='nndsvda')
    W_nmf = nmf.fit_transform(R_nmf_input)
    t_nmf = time.time() - t0
    nmf_models[K] = nmf
    print(f"✓ NMF (K={K:2d}, alpha_W={a:.2f}) fitted in {t_nmf:5.1f}s | Reconstruction Loss: {nmf.reconstruction_err_:.2f}")

nmf_best = nmf_models[15]
item_factors_nmf = nmf_best.components_.T
norms_nmf = np.linalg.norm(item_factors_nmf, axis=1, keepdims=True)
norms_nmf[norms_nmf == 0] = 1.0
item_factors_nmf_norm = item_factors_nmf / norms_nmf
print(f"✓ NMF Normalized Item Factors: {item_factors_nmf_norm.shape}")
```

## 14. Additional Model: FilmAura Tri-Hybrid Late-Fusion Ensemble
### Model Slot: Additional Model | Class: Custom Late-Fusion Ensemble
Blends normalized prediction score vectors from:
1. **Semantic Content Engine:** Sublinear TF-IDF over titles, genres, and community tags ($S_{content}$).
2. **Latent Matrix Factorization Engine:** SVD / NMF latent cosine similarity ($S_{latent}$).
3. **Affective Mood Distribution:** Continuous mood resonance score ($S_{mood}$).
4. **Bayesian Popularity Prior:** Weighted shrinked popularity rating ($S_{bayesian}$).

$$\text{Final Score}(i) = w_{content} S_{content}(i) + w_{mood} S_{mood}(i) + w_{latent} S_{latent}(i) + w_{pop} S_{bayesian}(i)$$
- Dynamic blend weights adapt when query includes an anchor film vs. cold-start mood discovery.


```python
# 1. Content Representation: Sublinear TF-IDF
print("Vectorizing semantic content metadata with TF-IDF...")
catalog_active['content_text'] = catalog_active.apply(
    lambda r: f"{r['clean_title']} {' '.join(r['genres_list'])} {' '.join(r['tags_list'])}",
    axis=1
)

tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=4000,
    ngram_range=(1, 2),
    sublinear_tf=True
)
tfidf_matrix = tfidf.fit_transform(catalog_active['content_text'])
print(f"✓ TF-IDF Sparse Feature Matrix: {tfidf_matrix.shape}")

# 2. Tri-Hybrid Late-Fusion Recommender Class
class FilmAuraTriHybridEnsemble:
    def __init__(self, catalog_df, tfidf_mat, svd_factors, baseline_mod):
        self.catalog = catalog_df.reset_index(drop=True)
        self.tfidf_matrix = tfidf_mat
        self.svd_factors = svd_factors
        self.baseline = baseline_mod
        self.mid_to_idx = {mid: i for i, mid in enumerate(self.catalog['movieId'])}
        
    def recommend(self, mood='Dark', genre=None, anchor_mid=None, top_k=5, weights=None):
        N = len(self.catalog)
        is_cold = (anchor_mid is None or anchor_mid not in self.mid_to_idx)
        
        # Default dynamic blend weights
        if weights is None:
            w_content, w_mood, w_latent, w_pop = (0.00, 0.45, 0.20, 0.35) if is_cold else (0.40, 0.30, 0.20, 0.10)
        else:
            w_content, w_mood, w_latent, w_pop = weights
            
        # 1. Content Similarity Vector
        if not is_cold:
            a_idx = self.mid_to_idx[anchor_mid]
            c_sim = cosine_similarity(self.tfidf_matrix[a_idx], self.tfidf_matrix)[0]
            l_sim = np.dot(self.svd_factors, self.svd_factors[a_idx])
        else:
            c_sim = np.full(N, 0.5)
            l_sim = np.full(N, 0.5)
            
        # 2. Mood Alignment Vector
        m_key = f"score_{mood.lower()}"
        m_sim = self.catalog[m_key].values if m_key in self.catalog.columns else np.full(N, 0.5)
        
        # 3. Bayesian Popularity Vector (Min-Max Scaled)
        b_scores = self.catalog['bayesian_score'].values
        b_norm = (b_scores - b_scores.min()) / (b_scores.max() - b_scores.min() + 1e-8)
        
        # Late-Fusion Combined Score
        hybrid_scores = (w_content * c_sim) + (w_mood * m_sim) + (w_latent * l_sim) + (w_pop * b_norm)
        
        # Filter candidates
        candidate_indices = np.arange(N)
        if genre and genre != 'All Genres':
            genre_mask = self.catalog['genres_list'].apply(lambda gl: genre in gl).values
            candidate_indices = candidate_indices[genre_mask]
            
        if not is_cold and a_idx in candidate_indices:
            candidate_indices = candidate_indices[candidate_indices != a_idx]
            
        sub_scores = hybrid_scores[candidate_indices]
        top_sub_idx = np.argsort(sub_scores)[::-1][:top_k]
        top_indices = candidate_indices[top_sub_idx]
        
        results = self.catalog.iloc[top_indices].copy()
        results['match_score'] = sub_scores[top_sub_idx]
        results['match_pct'] = np.clip(np.round(results['match_score'] * 100), 50, 99).astype(int)
        return results

ensemble_recommender = FilmAuraTriHybridEnsemble(catalog_active, tfidf_matrix, item_factors_svd_norm, baseline_best)
print("✓ FilmAura Tri-Hybrid Late-Fusion Ensemble initialized.")
```

## 15. Top-K Ranking Evaluation Suite
We evaluate all 5 candidate models on the held-out test interactions across standard Information Retrieval metrics:
- **Precision@K:** Fraction of top-$K$ recommendations that the user liked (rating $\ge 3.5$).
- **Recall@K:** Fraction of total relevant user items captured in top-$K$.
- **HitRate@K:** Probability of encountering at least one relevant item in top-$K$.
- **NDCG@K:** Normalized Discounted Cumulative Gain accounting for rank positions.


```python
def evaluate_ranking_models(sample_size=1000, K=10):
    print(f"Running Top-{K} Ranking Evaluation over {sample_size} test user profiles...")
    test_user_sample = pd.Series(test_eval['userId'].unique()).sample(n=sample_size, random_state=SEED).values
    
    metrics = {
        'Baseline (Bayesian Pop)': {'precision': [], 'recall': [], 'hit': [], 'ndcg': []},
        'Model 1 (Item-Item KNN)': {'precision': [], 'recall': [], 'hit': [], 'ndcg': []},
        'Model 2 (Matrix Factor SVD)': {'precision': [], 'recall': [], 'hit': [], 'ndcg': []},
        'Model 3 (Non-Negative NMF)': {'precision': [], 'recall': [], 'hit': [], 'ndcg': []},
        'FilmAura Tri-Hybrid Ensemble': {'precision': [], 'recall': [], 'hit': [], 'ndcg': []}
    }
    
    # Pre-compute top baseline recommendations
    top_baseline_ids = set(baseline_best.recommend(K)['movieId'].values)
    
    # Evaluation loop
    for uid in test_user_sample:
        u_train = train_eval[train_eval['userId'] == uid]
        u_test = test_eval[test_eval['userId'] == uid]
        
        ground_truth = set(u_test[u_test['rating'] >= 3.5]['movieId'].values)
        if not ground_truth:
            continue
            
        seed_mid = u_train.sort_values('rating', ascending=False)['movieId'].iloc[0] if not u_train.empty else None
        
        # Model predictions
        preds = {}
        preds['Baseline (Bayesian Pop)'] = list(top_baseline_ids)
        
        # SVD Top-K
        if seed_mid in movie_to_idx:
            s_idx = movie_to_idx[seed_mid]
            svd_scores = np.dot(item_factors_svd_norm, item_factors_svd_norm[s_idx])
            top_svd_idx = np.argsort(svd_scores)[::-1][1:K+1]
            preds['Model 2 (Matrix Factor SVD)'] = [catalog_active.iloc[i]['movieId'] for i in top_svd_idx]
            
            # NMF Top-K
            nmf_scores = np.dot(item_factors_nmf_norm, item_factors_nmf_norm[s_idx])
            top_nmf_idx = np.argsort(nmf_scores)[::-1][1:K+1]
            preds['Model 3 (Non-Negative NMF)'] = [catalog_active.iloc[i]['movieId'] for i in top_nmf_idx]
            
            # KNN Top-K
            _, knn_n = knn_best.kneighbors(R_item_user[s_idx], n_neighbors=K+1)
            preds['Model 1 (Item-Item KNN)'] = [catalog_active.iloc[i]['movieId'] for i in knn_n[0][1:]]
            
            # Ensemble Top-K
            ens_res = ensemble_recommender.recommend(anchor_mid=seed_mid, top_k=K)
            preds['FilmAura Tri-Hybrid Ensemble'] = list(ens_res['movieId'].values)
        else:
            preds['Model 2 (Matrix Factor SVD)'] = list(top_baseline_ids)
            preds['Model 3 (Non-Negative NMF)'] = list(top_baseline_ids)
            preds['Model 1 (Item-Item KNN)'] = list(top_baseline_ids)
            ens_res = ensemble_recommender.recommend(top_k=K)
            preds['FilmAura Tri-Hybrid Ensemble'] = list(ens_res['movieId'].values)
            
        # Metric Calculations
        for m_name, rec_list in preds.items():
            hits = [1 if m in ground_truth else 0 for m in rec_list[:K]]
            prec = sum(hits) / K
            rec = sum(hits) / len(ground_truth)
            hit = 1.0 if sum(hits) > 0 else 0.0
            
            # NDCG@K
            dcg = sum((2**h - 1) / np.log2(idx + 2) for idx, h in enumerate(hits))
            idcg = sum((2**1 - 1) / np.log2(idx + 2) for idx in range(min(len(ground_truth), K)))
            ndcg = (dcg / idcg) if idcg > 0 else 0.0
            
            metrics[m_name]['precision'].append(prec)
            metrics[m_name]['recall'].append(rec)
            metrics[m_name]['hit'].append(hit)
            metrics[m_name]['ndcg'].append(ndcg)
            
    summary = []
    for m_name, m_dict in metrics.items():
        summary.append({
            'Model Name': m_name,
            'Precision@10': np.mean(m_dict['precision']),
            'Recall@10': np.mean(m_dict['recall']),
            'HitRate@10': np.mean(m_dict['hit']),
            'NDCG@10': np.mean(m_dict['ndcg'])
        })
    return pd.DataFrame(summary)

eval_df = evaluate_ranking_models(sample_size=800, K=10)
print("\n" + "="*70)
print("BENCHMARK EVALUATION RESULTS ON HELD-OUT MOVIELENS 32M INTERACTIONS")
print("="*70)
display(eval_df.round(4))
```

## 18. Model Interpretability & Dynamic Rationale Generation
To eliminate black-box opacity, FilmAura generates dynamic natural language explanations for every surfaced recommendation.


```python
def generate_recommendation_rationale(row, mood, anchor_title=None):
    reasons = []
    reasons.append(f"Calibrated for your '{mood}' mood profile (score: {row[f'score_{mood.lower()}']:.2f})")
    if anchor_title:
        reasons.append(f"shares narrative and latent collaborative resonance with '{anchor_title}'")
    if row['rating_count'] >= 500:
        reasons.append(f"strongly endorsed by {row['rating_count']:,} community cinephiles ({row['rating_mean']}/5.0)")
    return "; ".join(reasons) + "."

demo_results = ensemble_recommender.recommend(mood='Dark', genre='Crime', anchor_mid=sample_match.iloc[0]['movieId'] if not sample_match.empty else None, top_k=3)

print("Demo Recommendations with Dynamic Interpretability:")
for idx, (_, film) in enumerate(demo_results.iterrows(), 1):
    rationale = generate_recommendation_rationale(film, 'Dark', sample_title)
    print(f"\n[{idx}] {film['clean_title']} ({film['year']}) — {film['match_pct']}% Match")
    print(f"    Genres: {', '.join(film['genres_list'])}")
    print(f"    Rationale: {rationale}")
```
